```python
import numpy as np
import matplotlib.pyplot as plt

# Increased dataset for Mount Fuji - 100 layers of geological data
layers = 100 # Increased to 100 layers
mean_diamonds = 500 # Mean number of diamonds per layer
std_dev = 150 # Standard deviation for diamond count

# Random diamond counts following a normal distribution
diamond_counts = np.random.normal(loc=mean_diamonds, scale=std_dev, size=layers).astype(int)
diamond_counts = np.clip(diamond_counts, 0, None) # Avoid negative counts

# Plotting the data
plt.figure(figsize=(12, 6))
plt.bar(range(1, layers+1), diamond_counts, color='c')
plt.xlabel("Geological layer")
plt.ylabel("Diamond count")
plt.title("Simulated Number of Diamonds on Mount Fuji")
plt.show()
```
